# Daikibo Telemetry Analysis

Identify which factory had the most machine breakdowns in May 2021 and which machine types failed most frequently at that factory.

Place your telemetry JSON at `data/daikibo_telemetry.json`.

```python
import pandas as pd
import json

# load data
with open('data/daikibo_telemetry.json', 'r') as file:
    data = json.load(file)

df = pd.json_normalize(data)

print(df.head())
print(df.columns)
print(df.info())

# filter to breakdowns / unhealthy statuses
breakdowns = df[df['status'] == 'unhealthy']
```

```python
# which factory had the most breakdowns
factory_breakdowns = (
    breakdowns.groupby('factory')
    .size()
    .sort_values(ascending=False)
)

print(factory_breakdowns)
```

```python
# top factory and machine types that broke down there
top_factory = factory_breakdowns.idxmax()

machine_breakdowns = (
    breakdowns[breakdowns['factory'] == top_factory]
    .groupby('machine')
    .size()
    .sort_values(ascending=False)
)

print('Factory with most breakdowns:', top_factory)
print(machine_breakdowns)
```
